```python
import os

# ============================================================
# 1. CREAR LA CARPETA utils
# ============================================================

os.makedirs("utils", exist_ok=True)


# ============================================================
# 2. CREAR __init__.py
# ============================================================

with open("utils/__init__.py", "w") as archivo:
    archivo.write("")


# ============================================================
# 3. CREAR ARCHIVO CON MAIN
# ============================================================

codigo_con_main = '''
def extraer():
    print("Extrayendo datos...")

def transformar():
    print("Transformando datos...")

def cargar():
    print("Cargando datos...")


if __name__ == "__main__":
    extraer()
    transformar()
    cargar()
'''

with open("utils/con_main.py", "w") as archivo:
    archivo.write(codigo_con_main)


# ============================================================
# 4. CREAR ARCHIVO SIN MAIN
# ============================================================

codigo_sin_main = '''
def extraer():
    print("Extrayendo datos...")

def transformar():
    print("Transformando datos...")

def cargar():
    print("Cargando datos...")


extraer()
transformar()
cargar()
'''

with open("utils/sin_main.py", "w") as archivo:
    archivo.write(codigo_sin_main)


# ============================================================
# 5. MOSTRAR LO QUE CREAMOS
# ============================================================

print("Archivos creados:")
print("- utils/")
print("  ├── __init__.py")
print("  ├── con_main.py")
print("  └── sin_main.py")
```
